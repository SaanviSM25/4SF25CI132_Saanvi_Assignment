Question 1:

```python
data={
    'a':[5,89,45,79,64],
    'b':[2,4,6,8,10],
    'c':[7,16,25,39,55]
}
df=pd.DataFrame(data)
print(df)
df['Product']=df['a']*df['b']
print(df)
```

Question 3:

```python
data={
    'a':[5,89,np.nan,79,64],
    'b':[2,4,6,8,np.nan],
    'c':[np.nan,16,25,39,55]
}
df=pd.DataFrame(data)
print(df)
df['a']=df['a'].fillna(df['a'].mean())
df['b']=df['b'].fillna(df['b'].mean())
df['c']=df['c'].fillna(df['c'].mean())
print(df)
```

Question 5:

```python
data={
    'Category':['A','B','C','B','C','A'],
    'Values':[4,6,7,25,42,32]
}
df=pd.DataFrame(data)
print(df)
grouped=df.groupby('Category')['Values']
print(grouped.sum())
print(grouped.mean())
```

Question 6:

```python
data={
    'Products':['White Nights','The Silent Patient','Laptop','Tablet','Chocolate','400 Days','Red Velvet'],
    'Category':['Books','Books','Electronics','Electronics','Cake','Books','Cake'],
    'Sales':[350,550,50000,30000,650,250,700]
}
df=pd.DataFrame(data)
grouped=df.groupby('Category')['Sales'].sum()
print(grouped)
```

Question 7:

```python
Students={
    'Id':[1,2,3,4],
    'Names':['Rahul','Priya','Ananya','Ram']
}
student_df=pd.DataFrame(Students)
print(student_df)
Marks={
    'Id':[1,2,3,4],
    'marks':[78,65,89,97]
}
mark_df=pd.DataFrame(Marks)
print(mark_df)
merged_df=pd.merge(student_df,mark_df,on='Id')
print(merged_df)
```

Question 9:

```python
data={
    'Category':['A','A','A','B','B','B'],
    'SubCategory':['X','Y','X','Y','X','Y'],
    'Value':[5,25,56,78,34,33]
}
df=pd.DataFrame(data)
print(df)
grouped=df.groupby(['Category','SubCategory'])['Value'].sum()
print(grouped)
```

Question 10:

```python
data={
    'X':[2,10,3,15,4,20],
    'Y':[98,65,32,78,46,23],
    'Z':[19,20,36,27,56,45]
}
df=pd.DataFrame(data)
print(df)
def double(n):
    return n*2
df=df.apply(double)
print(df)
```

Question 11:

```python
data={
    'X':[2,10,3,15,4,20],
    'Y':[98,65,32,78,46,23],
    'Z':[19,20,36,27,56,45]
}
df=pd.DataFrame(data)
print(df)
df['Sum']=df.apply(lambda x: x['X']+x['Y']+x['Z'],axis=1)
print(df)
```
